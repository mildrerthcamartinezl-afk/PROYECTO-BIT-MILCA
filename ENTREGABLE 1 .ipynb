# 3. Clasificación de estudiantes
Utilizando estructuras if, elif y else, clasifica a cada estudiante según su nota.
Nota    Clasificación
Mayor o igual a 4.5    Excelente
Entre 4.0 y 4.49    Bueno
Entre 3.0 y 3.99    Aceptable
Menor a 3.0    Reprobó
Imprime el nombre del estudiante junto con su clasificación.

```python
for nombre, edad, nota in ESTUDIANTES:
    if nota >=4.5:
      print(f"{nombre} tiene {edad}  años y obtuvo una nota de {nota} -> Excelente")
    elif nota >=4.0 and nota <=4.49:
      print(f"{nombre} tiene {edad} años y obtuvo una nota de {nota} -> Bueno")
    elif nota >=3.0 and nota <=3.99:
     print(F"{nombre} tiene {edad} años y obtuvo una nota de {nota} -> Aceptable")
    else:
      print(f"{nombre} tiene {edad} años y obtuvo una nota de {nota} -> Reprobo")
```

# 5. Búsqueda de estudiante
Solicita al usuario el nombre de un estudiante mediante input().
Si el estudiante existe en la lista, imprime un mensaje indicando que fue encontrado.
Si no existe, informa que no se encontró ningún estudiante con ese nombre.


```python
nombre_buscar = input("Ingrese el nombre del estudiante a buscar: ")
encontrado = False

for nombre, edad, nota in ESTUDIANTES:
    if nombre.lower() == nombre_buscar.lower():
        print(f"{nombre} fue encontrado y tiene {edad} años y la nota que obtuvo fue {nota}.")
        encontrado = True
        break

if not encontrado:
    print("No se encontró ningún estudiante con ese nombre.")
```

#7. Cantidad de estudiantes por ciudad
A partir del diccionario anterior, construye un nuevo diccionario que indique cuántos estudiantes viven en cada ciudad.
Ejemplo:

{
    "Bogotá": 3,
    "Medellín": 2,
    "Pereira": 1
}

```python
ciudades ={ "MILDRETH": "BOGOTA", "CAROLINA": "BARRANQUILLA", "MARGI": "BARRANQUILLA", "CONSUELO": "BOGOTA", "TATIANA": "BOGOTA", "JHON":"CARTAGENA", "HOLLMAN": "MOSQUERA"}
conteo_ciudades = {}
for nombre in ciudades:
  ciudad = ciudades [nombre]
  if ciudad in conteo_ciudades:
    conteo_ciudades[ciudad] += 1
  else:
    conteo_ciudades[ciudad] =1
print(conteo_ciudades)
```

# 8. Ciclo While
Construye un programa que permita ingresar números utilizando un ciclo while.
El programa debe finalizar únicamente cuando el usuario ingrese el número 0.
Al finalizar muestra:
    • Cantidad de números ingresados (sin contar el 0).
    • Suma total.
    • Promedio de los números ingresados.


```python
suma = 0
contador = 0

numero = float(input("Ingrese un número (0 para finalizar): "))

while numero != 0:
    suma += numero
    contador += 1
    numero = float(input("Ingrese otro número (0 para finalizar): "))

if contador > 0:
    promedio = suma / contador
    print("Cantidad de números ingresados:", contador)
    print("Suma total:", suma)
    print("Promedio:", promedio)
else:
    print("No se ingresaron números.")
```

# 9. Uso de break y continue
Recorre los números del 1 al 30.
    • Si el número es múltiplo de 3, utiliza continue.
    • Si el número es igual a 25, utiliza break.
Imprime únicamente los números que fueron procesados por el ciclo.

```python
for numero in range(1, 31):
    if numero % 3 == 0:
        continue
    if numero == 25:
        break
    print(numero)
```
