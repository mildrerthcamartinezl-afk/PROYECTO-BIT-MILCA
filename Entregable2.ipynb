```python
#2. Limpieza de datos
#Realiza las transformaciones necesarias para obtener un DataFrame consistente. Debes:
#2.1  Identificar las columnas con valores faltantes y aplica una estrategia de tratamiento adecuada.
#2.2 Detectar y elimina los registros duplicados cuando corresponda.
#2.3 Convertir las columnas que tengan tipos de datos incorrectos.
#2.4 Identificar al menos dos variables categóricas que presenten inconsistencias y estandarízalas.
#2.5 Después de realizar la limpieza, demuestra mediante código que las transformaciones fueron aplicadas correctamente.
```

```python
#2.1  Identificar las columnas con valores faltantes y aplica una estrategia de tratamiento adecuada.

print(df.isnull().sum())
df['Weight kg'] = df['Weight kg'].fillna(df['Weight kg'].mean())
df['Body Length cm'] = df['Body Length cm'].fillna(df['Body Length cm'].mean())
df['Animal name'] = df['Animal name'].fillna("Unknown")
df['Country'] = df['Country'].fillna("Unknown")
```

```python
#2.2 Detectar y elimina los registros duplicados cuando corresponda.

print("Duplicados antes:", df.duplicated().sum())
df = df.drop_duplicates()
print("Duplicados después:", df.duplicated().sum())
```

```python
#2.3 Convertir las columnas que tengan tipos de datos incorrectos.
# Convertir columnas numéricas que estaban como texto


df["Weight kg"] = pd.to_numeric(df["Weight kg"], errors="coerce")
df["Body Length cm"] = pd.to_numeric(df["Body Length cm"], errors="coerce")
df["Latitude"] = pd.to_numeric(df["Latitude"], errors="coerce")
df["Longitude"] = pd.to_numeric(df["Longitude"], errors="coerce")
print(df.dtypes)
```

```python
#2.4 Identificar al menos dos variables categóricas que presenten inconsistencias y estandarízalas.

# Normalizar nombres de países
df['Country'] = df['Country'].str.strip().str.lower()
df['Country'] = df['Country'].replace({
    'pl': 'poland', 'poland': 'poland',
    'hungry': 'hungary', 'hu': 'hungary',
    'cz': 'czech republic', 'czech': 'czech republic',
    'de': 'germany'
})

# Normalizar tipos de animales
df['Animal type'] = df['Animal type'].str.strip().str.lower()
df['Animal type'] = df['Animal type'].replace({
    'red squirel': 'red squirrel',
    'red squerrel': 'red squirrel',
    'european bisson': 'european bison',
    'european bison*': 'european bison',
    'hedgehod': 'hedgehog',
    'wedghed': 'hedgehog'
})

print(df["Country"].unique())
print(df["Animal type"].unique())
```

```python
#2.5 Después de realizar la limpieza, demuestra mediante código que las transformaciones fueron aplicadas correctamente.

print("Valores faltantes por columna:\n", df.isnull().sum())
print("\nRegistros duplicados:", df.duplicated().sum())
print("\nTipos de datos por columna:\n", df.dtypes)
print("\nPaíses únicos:\n", df["Country"].unique())
print("\nTipos de animales únicos:\n", df["Animal type"].unique())
```

```python
#3.1 Crea una variable utilizando: np.where() La variable debe representar una condición relevante para el contexto de los datos.

import numpy as np
promedio_peso = np.nanmean(df["Weight kg"])
df["Peso_categoria"] = np.where(df["Weight kg"] > promedio_peso, "Superior al promedio", "Inferior al promedio")
print(df[["Weight kg", "Peso_categoria"]].head())
```

```python
#4.1 Debes realizar al menos 4 análisis diferentes.

# 4.1.1 Número de observaciones por país
observaciones_pais = df.groupby("Country")["Country"].count()
print(observaciones_pais)

# 4.1.2 Promedio del peso por género
promedio_peso_genero = df.groupby("Gender")["Weight kg"].mean()
print(promedio_peso_genero)

# 4.1.3 Promedio de la longitud corporal por tipo de animal
promedio_longitud_animales = df.groupby("Animal type")["Body Length cm"].mean()
print(promedio_longitud_animales)

# 4.1.4 Promedio del peso por género y país
promedio_peso_genero_pais = df.groupby(["Country", "Gender"])["Weight kg"].mean()
print(promedio_peso_genero_pais)
```

```python
#4.2 Cuál es el número de observaciones por tipo de animal?
# 4.2 Número de observaciones por tipo de animal
observaciones_animales = df.groupby("Animal type")["Animal type"].count()
print(observaciones_animales)
```
