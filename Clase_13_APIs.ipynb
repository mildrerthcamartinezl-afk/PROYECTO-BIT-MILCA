## 1. Qué es una API?

**API** significa *Application Programming Interface*.

Es una interfaz que permite que diferentes aplicaciones o sistemas **se comuniquen entre sí**.

### Ejemplos cotidianos

| Aplicación | ¿Qué podría obtener mediante una API? |
|---|---|
| App del clima | Temperatura, lluvia, pronóstico |
| Spotify | Canciones, artistas, álbumes |
| Uber | Viajes, ubicaciones, conductores |
| Redes sociales | Publicaciones, perfiles, comentarios |

**Idea clave:** una API funciona como un intermediario entre aplicaciones.

## 2. Qué ocurre cuando consumimos una API?

El flujo básico es:

```text
 Python
   │
   │  Request (solicitud)
   ▼
  API
   │
   │  Response (respuesta)
   ▼
 Python
```

### Conceptos importantes

| Concepto | Significado |
|---|---|
| **Request** | Solicitud que hacemos a la API |
| **Response** | Respuesta que recibimos |
| **Endpoint** | URL específica de la API |
| **GET** | Solicitar/obtener información |
| **JSON** | Formato frecuente para intercambiar datos |
| **Status code** | Código que indica cómo salió la petición |

## 3. Códigos de respuesta HTTP

| Código | Significado |
|---:|---|
| **200** | OK — la petición fue exitosa |
| **201** | Created — se creó un recurso |
| **400** | Bad Request — solicitud incorrecta |
| **401** | Unauthorized — requiere autorización |
| **404** | Not Found — recurso no encontrado |
| **500** | Internal Server Error — error del servidor |


## 4. Librerías

````python
import requests
import pandas as pd
````

## 5. Primera petición HTTP

Vamos a consumir una API pública de prueba: **JSONPlaceholder**.

Endpoint:

`https://jsonplaceholder.typicode.com/users`

````python
url = "https://jsonplaceholder.typicode.com/users"

respuesta = requests.get(url)

print(respuesta)
````

### Qué recibimos?

Cuando ejecutamos `requests.get()`, obtenemos un objeto **Response**.

Ahora podemos revisar su código de estado:

````python
print(respuesta.status_code)
````

## 6. Convertir la respuesta a JSON

La respuesta puede contener datos en formato JSON.

Con `.json()` convertimos esa respuesta a una estructura que Python puede trabajar.

````python
datos = respuesta.json()

print(datos)
````

### Qué estructura tenemos?

Una respuesta como esta suele tener la estructura:

```text
lista
 ├── diccionario (usuario 1)
 ├── diccionario (usuario 2)
 ├── diccionario (usuario 3)
 └── ...
```

## 7. Obtener información específica

Podemos acceder a un elemento de la lista y luego consultar una de sus claves.

````python
print(datos[0]["name"])
print(datos[0]["email"])
````

### La lógica

```text
datos[0]              → primer usuario
datos[0]["name"]      → nombre del primer usuario
datos[0]["email"]     → email del primer usuario
```

## 8. Recorrer todos los usuarios

Si queremos trabajar con todos los elementos de la lista, podemos utilizar un ciclo `for`.

````python
for usuario in datos:
    print(usuario["name"])
    print(usuario["email"])
    print("---")
````

## 9. Información anidada

Los datos de una API pueden estar dentro de otros datos.

Por ejemplo:

```text
usuario
 ├── name
 ├── email
 └── address
       ├── street
       ├── city
       └── zipcode
```

Para llegar a la ciudad:

````python
print(datos[0]["address"]["city"])
````

### Nota:

Cada `[]` nos permite **entrar un nivel más** en la estructura de los datos.

## 10. De API a DataFrame

Una vez tenemos los datos en Python, podemos llevarlos a Pandas para analizarlos.

````python
df = pd.DataFrame(datos)

df.head()
````

### Explorar los datos

````python
df.shape
````

````python
df.columns
````

````python
df["city"] = df["address"].apply(lambda x: x["city"])

df[["name", "email", "city"]]
````
